7. String Construction

```python
first_name = input("Enter your first name: ").strip()
last_name = input("Enter your last name: ").strip()

full_name = first_name + " " + last_name

print("Normal Form:", full_name)
print("Uppercase:", full_name.upper())
print("Lowercase:", full_name.lower())
```

8. Method Selection Challenge

```python
text = input("Enter a string: ")

print("Length:", len(text))
print("First Character:", text[0])
print("Last Character:", text[-1])
print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
print("Title Case:", text.title())
print("Reversed:", text[::-1])
```

9.operator Logic

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)

if b != 0:
    print("Division:", a / b)
    print("Floor Division:", a // b)
    print("Modulus:", a % b)
else:
    print("Division: Not possible")
    print("Floor Division: Not possible")
    print("Modulus: Not possible")

print("Exponentiation:", a ** b)
print("First number greater than second:", a > b)
print("Both numbers are equal:", a == b)

if b != 0 and a % b == 0:
    print("Divisible: Yes")
else:
    print("Divisible: No")
```

10. Conditional Classification

```python
marks = int(input("Enter student's marks: "))

if marks < 0 or marks > 100:
    print("Invalid Input")
elif marks >= 90:
    print("Excellent")
elif marks >= 75:
    print("Very Good")
elif marks >= 60:
    print("Good")
elif marks >= 40:
    print("Pass")
else:
    print("Fail")
```

11. Nested Conditional Challenge

```python
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin":
    if password == "12345":
        print("Login Successful")
    else:
        print("Username correct but password is wrong")
else:
    print("Wrong username")
```

13. while Loop Counter

```python
i = 1
count = 0

while i <= 20:
    if i % 3 == 0:
        print(i)
        count = count + 1

    i = i + 1

print("Total numbers divisible by 3:", count)
```
